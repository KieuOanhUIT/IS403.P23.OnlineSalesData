```python
import pandas as pd

# 1. Đọc file gốc
df = pd.read_csv("cinemaTicket_Ref.csv")

# 2. Chuyển 'date' về kiểu datetime
df['date'] = pd.to_datetime(df['date'])

# 3. Tổng hợp theo ngày
daily_df = df.groupby('date').agg({
    'total_sales': 'sum',
    'tickets_sold': 'sum',
    'ticket_use': 'sum',
    'occu_perc': 'mean',
    'film_code': 'nunique',
    'cinema_code': 'nunique'
}).reset_index()

# 4. Đổi tên cột
daily_df.rename(columns={
    'total_sales': 'daily_sales',
    'tickets_sold': 'daily_tickets',
    'ticket_use': 'daily_used_tickets',
    'occu_perc': 'avg_occupancy',
    'film_code': 'unique_films',
    'cinema_code': 'unique_cinemas'
}, inplace=True)

# ➕ Thêm lại các cột thời gian
daily_df['day'] = daily_df['date'].dt.day
daily_df['month'] = daily_df['date'].dt.month
daily_df['quarter'] = daily_df['date'].dt.quarter
daily_df['year'] = daily_df['date'].dt.year

# 5. Sắp xếp theo ngày và lưu
daily_df = daily_df.sort_values('date')
daily_df.to_csv("cinema_daily_sales.csv", index=False)
```
